```python
# [AI Disclaimer example: Used ChatGPT with prompt “teach me 'INT_MAX / 10 and INT_MIN / 10' in python”]
class Solution(object):
    def reverse(self, x):
        """
        :type x: int
        :rtype: int
        """
        INT_MIN, INT_MAX = -2**31, 2**31 - 1
        neg = False
        if x < 0:
          neg = True
          x = -x
        ans = 0
        while x != 0:
          sbzuoye = x %10
          ans = ans * 10 + sbzuoye
          x //= 10
        if neg == True:
          x = -x
        if ans < INT_MIN or ans > INT_MAX:
            return 0
        return ans


```

https://leetcode.com/problems/valid-palindrome/solutions/7197820/clean-simple-efficient-beginner-friendly-easy-explanation

After reviewing the tab, I learnd that there is a thing called [::-1] which flips the list, and I learned isalnum() to see if its a letter. I also learned to use lower() which turns uppercase to lowercase.


```python
# [AI Disclaimer example: Used ChatGPT with prompt “What does salnum() and lower() mean in python”]
class Solution(object):
    def isPalindrome(self, s):
        """
        :type s: str
        :rtype: bool
        """
        NewS = ""
        for c in s:
            if c.isalnum():
                NewS += c.lower()
        if NewS == NewS[::-1]:
            return True
        else:
            return False
```
